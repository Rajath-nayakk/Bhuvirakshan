# Forest & River Guardian: Exploratory Data Analysis
This notebook explores multi-temporal satellite imagery (Sentinel-2 and Sentinel-1 SAR),
acoustic audio samples, and camera-trap frames.

```python
import rasterio
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# Inspect sample satellite tile
tile_path = Path('../data/samples/satellite/tile_000_2023_before.tif')
if tile_path.exists():
    with rasterio.open(tile_path) as src:
        print('Tile CRS:', src.crs)
        print('Band Count:', src.count)
        print('Dimensions:', src.shape)
        b4 = src.read(3)
        b8 = src.read(4)
        ndvi = (b8 - b4) / (b8 + b4 + 1e-6)
        
    plt.figure(figsize=(6, 5))
    plt.imshow(ndvi, cmap='RdYlGn')
    plt.title('NDVI Vegetation Index')
    plt.colorbar(label='NDVI')
    plt.show()
```

```python
# Inspect Audio Spectrogram
import soundfile as sf
import torch
from src.data.audio_data import AudioFeatureExtractor

wav_path = Path('../data/samples/audio/gunshot/gunshot_000.wav')
if wav_path.exists():
    data, sr = sf.read(str(wav_path))
    extractor = AudioFeatureExtractor()
    spec = extractor.extract_log_mel(torch.from_numpy(data).float())
    print('Spectrogram Shape:', spec.shape)
    
    plt.figure(figsize=(7, 3))
    plt.imshow(spec.squeeze().numpy(), origin='lower', aspect='auto', cmap='magma')
    plt.title('Gunshot Audio Spectrogram')
    plt.colorbar(label='Log Mel Power')
    plt.show()
```
