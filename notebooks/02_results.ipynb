# Forest & River Guardian: Results, Metrics & Visualizations
Analysis of change-detection models, acoustic classifier robustness, and 500m risk fusion.

```python
import json
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# Load Evaluation Metrics
metrics_dir = Path('../reports/metrics')
with open(metrics_dir / 'change_detection_metrics.json') as f:
    change_metrics = json.load(f)
with open(metrics_dir / 'audio_metrics.json') as f:
    audio_metrics = json.load(f)
with open(metrics_dir / 'camera_metrics.json') as f:
    camera_metrics = json.load(f)

print('=== Change Detection Comparison ===')
print(pd.DataFrame(change_metrics['model_comparison']).T)

print('\n=== Camera Trap mAP@0.50 ===', camera_metrics['mAP50'])
print('=== Audio Classifier Accuracy ===', audio_metrics['best_accuracy'])
```

```python
# Plot SNR Robustness Curve
snr_data = audio_metrics['snr_robustness_curve']
snrs = [-10, -5, 0, 5, 10, 15, 20]
accs = [snr_data[f'snr_{s}_db'] for s in snrs]

plt.figure(figsize=(7, 4))
plt.plot(snrs, [a * 100 for a in accs], marker='o', linewidth=2, color='#10b981')
plt.title('Audio Classifier Noise Robustness Across SNR (dB)')
plt.xlabel('Signal-to-Noise Ratio (dB)')
plt.ylabel('Classification Accuracy (%)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.ylim(30, 105)
plt.show()
```
